```python
# print name n times
def nameN(count,n):
    if count == n:
        return
    print("Ram",end=' ')
    count += 1
    nameN(count,n)

nameN(0,5)
    
```

```python
def numP(start,n):
    if start == n:
        return
    print(start,end=' ')
    numP(start+1,n)
numP(0,5)
```

```python
def numP(n):
    if n < 1:
        return
    print(n,end=' ')
    numP(n-1)
numP(5)
```

```python
def numP(n):
    if n == 1:
        return 1
    return n + numP(n-1)

print(numP(5))
```

```python
def factN(n):
    if n < 1:
        return 1
    return n * factN(n-1)

print(factN(5))
```

```python
def revA(arr):
    l = len(arr) -1
    arr2 = []
    if l == 0:
        return arr2
    arr2 = [l]
    l -= 1
    revA(arr)    

arr = [1,2,3,4,5]
revA(arr)
```

```python
def revA(arr,l,arr2):
    if l < 0:
        return arr2
    arr2.append(arr[l])
    return revA(arr,l-1,arr2)

arr = [1,2,3,4,5]
l = len(arr)-1
arr2 = []

p = revA(arr,l,arr2)
print(p)
```
