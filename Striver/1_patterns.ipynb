Given n, print a solid square of * of size n.

```python
def solidSqr(n):
    for i in range(n):
        for j in range(n):
            print("* ", end=' ')
        print()

n = int(input())
solidSqr(n)
```

Pattern 2 – Right-Angled Triangle (Stars)

```python
def rightPattern(n):
    for i in range(n):
        for j in range(n):
            if i >= j:
                print("* ", end=' ')
            else:
                break
        print()
        
n = int(input())
rightPattern(n)
```

Pattern 3 – Right-Angled Triangle (Numbers)

```python
def  rightNum(n):
    for i in range(n):
        for j in range(n):
            if i >= j:
                print(j+1,end=' ')
            else:
                break
        print()
        
n = int(input())
rightNum(n)
```

Pattern 4 – Increasing Numbers

```python
def rightRow(n):
    for i in range(n):
        for j in range(i+1):
            print(i+1,end=' ')
        print()
n = int(input())
rightRow(n)
```

Pattern 5 – Inverted Right-Angled Triangle

```python
def reverseRight(n):
    for i in range(n):
        for j in range(n-i):
            print("*",end=' ')
        print()
        
n = int(input())
reverseRight(n)
```

Pattern 6 – Inverted Number Triangle

```python
def reverseRightNum(n):
    for i in range(n):
        for j in range(n-i):
            print(j+1, end=' ')
        print()
        
n = int(input())
reverseRightNum(n)
```

Pattern 7 – Pyramid (Star Pyramid)

```python
n = 4
m = 7
mid = 3
z = 1
for i in range(n):
    for j in range(m):
        if i+j >= mid:
            if i+j <= z+2:
                print(" * ",end=" ")  
            else:
                print("   ",end=" ")       
        else:
            print("   ",end=" ")
    z+=2
    print()
```
