```python
n1 = 9
n2 = 12
arr1 = []
arr2 = []
result = []

for i in range(1,n1+1):
    if n1 % i == 0:  
        arr1.append(i) 
        
for i in range(1,n2+1):
    if n2 % i == 0:  
        arr2.append(i)  
# print(arr1)
# print(arr2)

for i in arr1:
    for j in arr2:
        if i == j:
            result.append(i)

# print(result)
print(f'GCD =',result[-1])
```

```python
a = 123  # 153 is armstrong
s = str(a)
count = 0 
total = 0 
for i in range(len(s)):
    count+=1
    
# print(count)
armstrong = list(map(int,s))
# print(armstrong)
for i in armstrong:
    z = i ** count 
    total += z
if total == a :
    print("Armstrong")
else:
    print('Not Armstrong')
```
