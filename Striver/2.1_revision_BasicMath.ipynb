```python
n = 123
rev = 0 
while n > 0:
    d = n % 10
    rev = rev * 10 + d
    n = n // 10
print(rev)
```

```python
n = 12321 ; k = n
rev = 0 
while n > 0:
    d = n % 10
    rev = rev * 10 + d
    n = n // 10

if rev == k:
    print('Palindrome')
else:
    print('Not Palindrome')
```

```python
n = 16
m = 32
while n % m != 0:
    temp = n % m 
    n = m
    m = temp

print(m)
```

```python
n = 153; s = str(n) ; l = len(s) ; total = 0
nums = list(map(int,s))
for i in nums:
    d = i ** l
    total += d
if total == n:
    print("Armstrong")
else:
    print("Not Armstrong")
```
