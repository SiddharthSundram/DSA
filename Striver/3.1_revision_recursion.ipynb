```python
#  1 to n times:-

def ntimes(start,n):
    if start > n:
        return 
    print(start,end=' ')
    ntimes(start + 1,n)
    
ntimes(1,5)
```

```python
# anything n-times:-

def ntimes(n):
    if n == 0:
        return 
    print(n,end=' ')
    ntimes(n-1)
    
ntimes(5)
```

```python
# sum n-times:-
def ntimes(n,s=0):
    if n == 0:
        return s
    s += n
    return ntimes(n-1,s)

print(ntimes(5))
```

```python
# sum n-times:-
def ntimes(n,s=1):
    if n == 0:
        return s
    s *= n
    return ntimes(n-1,s)

print(ntimes(5))
```

```python
def arrRev(arr,arr1,i):
    if i == 0:
        return arr1
    arr1.append(arr[i-1])
    return arrRev(arr,arr1,i-1)

arr = [1,2,3,4,5]
arr1 = []
i = len(arr)
arrRev(arr,arr1,i)
```

```python
def revArr2(arr):
    if not arr:
        return []
    return [arr[-1]] + revArr2(arr[:-1])

arr = [1,2,3,4,5]
revArr2(arr)
```

```python
def palin(s,i=0):
    if i >= len(s)//2:
        return True
    if s[i] != s[len(s)-1-i]:
        return False
    return palin(s,i+1)
    

if palin('sidis'):
    print("Palindrome")
else:
    print("Not Palindrome")
```

```python
# factorial 
def fib(n):
    if n <= 1:
        return n
    a = fib(n-1)
    b = fib(n-2)
    return a + b

# fib(4)
for i in range(5):
    print(fib(i),end=' ')
```
